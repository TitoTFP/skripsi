# BAB 4 Generator Python

Notebook ini hanya melakukan orchestration: import fungsi generator dari `bab4/sections/*.py`, memanggilnya per subbab, lalu menulis manifest validasi ke `bab4/outputs/`.

```python
%load_ext autoreload
%autoreload 2

from pathlib import Path
import sys

ROOT = Path.cwd().resolve()
while not (ROOT / "pyproject.toml").exists() and ROOT.parent != ROOT:
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from bab4.artifacts import validate_manifest
from bab4.config import Bab4Config
from bab4.run_all import _write_manifest, _write_validation_report
from bab4.sections.s4_1_1 import generate_4_1_1
from bab4.sections.s4_1_2 import generate_4_1_2
from bab4.sections.s4_2 import generate_4_2
from bab4.sections.s4_3 import generate_4_3
from bab4.sections.s4_4 import generate_4_4_1, generate_4_4_2, generate_4_4_3
from bab4.sections.s4_5_6 import generate_4_5, generate_4_6
from bab4.sections.s4_7_8_9 import generate_4_7, generate_4_8, generate_4_9

def show(rows, limit=20):
    rows = list(rows)
    for row in rows[:limit]:
        print(row)
    if len(rows) > limit:
        print(f"... {len(rows) - limit} more rows")

def show_result(result):
    show(artifact.to_record() for artifact in result.artifacts)

config = Bab4Config.from_repo(ROOT)
if config.clean_outputs:
    config.reset_output_dirs()
else:
    config.ensure_output_dirs()
results = []
print(f"Output root: {config.output_root}")
```
